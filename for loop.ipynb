# for loop 

A for loop is used to repeat a block of code for each item in a sequence.

# syntax

for variable in sequence:

#code

```python
for i in range (1,11):
     print(i)
```

23. find the factorial of a number

```python
n=int(input("Enter a number: "))
fact=1

for i in range(1,n+1):   # factorial means multiplying all numbers from 1 to n (ex:- 5!,8!)
    fact=fact*i          # 1x2x3x4x5=120

print(fact)
    
```

Loop    i   fact = fact ×   	New fact
Start	-	       1	           1
1	   1	      1 × 1	           1
2	   2         1 × 2	           2
3      3	      2 × 3	           6
4	   4	      6 × 4	           24
5	   5	      24 × 5	       120

Enter a number: 5

fact = 1

i = 1 → fact = 1 × 1 = 1
i = 2 → fact = 1 × 2 = 2
i = 3 → fact = 2 × 3 = 6
i = 4 → fact = 6 × 4 = 24
i = 5 → fact = 24 × 5 = 120
